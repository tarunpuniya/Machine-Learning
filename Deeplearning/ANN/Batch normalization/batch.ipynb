# Batch normalization in Neural network

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('concertriccir2.csv')
df.head()
```

```python
X = df.iloc[:, 0:2]   # first two columns
y = df.iloc[:, 2]     # third column
```

```python
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=y)
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense , Dropout , BatchNormalization
```

```python
model1 = Sequential()

model1.add(Dense(2,activation='relu',input_dim=2))
model1.add(Dense(2,activation='relu'))
model1.add(Dense(1,activation='sigmoid'))
```

```python
model1.summary()
```

```python
model1.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history1 = model1.fit(X,y,epochs=200,validation_split=0.2)
```

```python
#Model training with batch normalization 

model2 = Sequential()

model2.add(Dense(2,activation='relu',input_dim=2))
model2.add(BatchNormalization())
model2.add(Dense(2,activation='relu'))
model2.add(BatchNormalization())
model2.add(Dense(1,activation='sigmoid'))
```

```python
model2.summary()
```

```python
model2.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history2 = model2.fit(X,y,epochs=200,validation_split=0.2)
```

```python
plt.plot(history1.history['val_accuracy'],color ='red')
plt.plot(history2.history['val_accuracy'],color ='green')
```

```python

def plot_decision_boundary(model, X, y, title):

    x_min, x_max = X.iloc[:, 0].min() - 1, X.iloc[:, 0].max() + 1
    y_min, y_max = X.iloc[:, 1].min() - 1, X.iloc[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 300),
        np.linspace(y_min, y_max, 300)
    )

    grid = np.c_[xx.ravel(), yy.ravel()]

    Z = model.predict(grid, verbose=0)
    Z = (Z > 0.5).astype(int)
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(7, 6))

    plt.contourf(xx, yy, Z, alpha=0.25)

    plt.scatter(
        X.iloc[:, 0],
        X.iloc[:, 1],
        c=y,
        edgecolors='black'
    )

    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.title(title)

    plt.show()
```

```python
plot_decision_boundary(
    model1,
    X,
    y,
    "Decision Boundary - Without Batch Normalization"
)

plot_decision_boundary(
    model2,
    X,
    y,
    "Decision Boundary - With Batch Normalization"
)
```

```python
def compare_decision_boundaries(model1, model2, X, y):

    x_min, x_max = X.iloc[:, 0].min() - 1, X.iloc[:, 0].max() + 1
    y_min, y_max = X.iloc[:, 1].min() - 1, X.iloc[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 300),
        np.linspace(y_min, y_max, 300)
    )

    grid = np.c_[xx.ravel(), yy.ravel()]

    Z1 = model1.predict(grid, verbose=0)
    Z1 = (Z1 > 0.5).astype(int).reshape(xx.shape)

    Z2 = model2.predict(grid, verbose=0)
    Z2 = (Z2 > 0.5).astype(int).reshape(xx.shape)

    fig, ax = plt.subplots(1, 2, figsize=(14, 6))

    ax[0].contourf(xx, yy, Z1, alpha=0.25)
    ax[0].scatter(
        X.iloc[:, 0],
        X.iloc[:, 1],
        c=y,
        edgecolors='black'
    )
    ax[0].set_title("Without Batch Normalization")
    ax[0].set_xlabel("Feature 1")
    ax[0].set_ylabel("Feature 2")

    ax[1].contourf(xx, yy, Z2, alpha=0.25)
    ax[1].scatter(
        X.iloc[:, 0],
        X.iloc[:, 1],
        c=y,
        edgecolors='black'
    )
    ax[1].set_title("With Batch Normalization")
    ax[1].set_xlabel("Feature 1")
    ax[1].set_ylabel("Feature 2")

    plt.tight_layout()
    plt.show()
```

```python
compare_decision_boundaries(model1, model2, X, y)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history1.history['loss'], label='Without BatchNorm')
plt.plot(history2.history['loss'], label='With BatchNorm')

plt.xlabel('Epoch')
plt.ylabel('Training Loss')
plt.title('Training Loss Comparison')
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history1.history['val_loss'], label='Without BatchNorm')
plt.plot(history2.history['val_loss'], label='With BatchNorm')

plt.xlabel('Epoch')
plt.ylabel('Validation Loss')
plt.title('Validation Loss Comparison')
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history1.history['accuracy'], label='Without BatchNorm')
plt.plot(history2.history['accuracy'], label='With BatchNorm')

plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Training Accuracy Comparison')
plt.legend()
plt.grid(True)

plt.show()
```
