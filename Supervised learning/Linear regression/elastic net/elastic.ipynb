# Elastic net regression 

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression , Lasso , Ridge , ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
model = LinearRegression()
model.fit(X_train,y_train)

y_pred = model.predict(X_test)
print("R2 score for test",r2_score(y_test,y_pred))

y_pred1 = model.predict(X_train)
print("R2 score for train",r2_score(y_train,y_pred1))
```

```python
ridge = Ridge(alpha=0.01)
ridge.fit(X_train,y_train)

rig_pred = ridge.predict(X_test)
print("R2 score for test: ",r2_score(y_test,rig_pred))

rig_pred1 = ridge.predict(X_train)
print("R2 score for train: ",r2_score(y_train,rig_pred1))
```
